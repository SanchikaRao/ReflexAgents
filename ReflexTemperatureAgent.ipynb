```python
class ReflexTmperatureAgent:
  def act(self,temp):
    if temp>30:
      return "Turn ON Fan"
    elif temp<20:
      return "Turn ON Heater"
    else:
      return "Maintain Current Temperature"
```

```python
agent=ReflexTmperatureAgent()
```

```python
temp=[10,30,45,15,40]
```

```python
for t in temp:
  action=agent.act(t)
  print("Temperature:",t)
  print("Action",action)
```
