```python
class AutomaticDoorAgent:
  def act(self,person):
    if person == True:
      return "Open the door"
    else:
      return "Close the door"
```

```python
agent=AutomaticDoorAgent()
```

```python
person = [True,False,False,True,True]
```

```python
for t in person:
  action=agent.act(t)
  print("Person Detected : ",t)
  print("Action : ",action)
```
