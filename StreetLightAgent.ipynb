```python
class StreetLightAgent:
  def act(self,light):
    if light == 'dark':
      return "Turn ON Street Light"
    else:
      return "Turn OFF Street Light"
```

```python
agent=StreetLightAgent()
```

```python
light=['dark','light','light','dark']
```

```python
for t in light:
  action=agent.act(t)
  print("Light : ",t)
  print("Action : ",action)
```
